```python
import pandas as pd
url="https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
titanic =pd.read_csv(url)
print(titanic.head())
```

```python
print(titanic.shape)
print(titanic.columns)
print(titanic.isnull().sum())
```

```python
titanic["age_group"] = pd.cut(titanic["age"], bins=[0, 12, 18, 35, 60, 100],
                                labels=["Child", "Teen", "Adult", "Middle-aged", "Senior"])
titanic.groupby("age_group")["survived"].mean()
```

```python
titanic.groupby(["pclass", "sex"])["survived"].mean()
```

```python
class_survival = titanic.groupby("pclass")["survived"].mean()
sex_survival = titanic.groupby("sex")["survived"].mean()

summary = f"""
Key Insight Summary:
- Passengers in 1st class had a {class_survival[1]*100:.1f}% survival rate,
  compared to just {class_survival[3]*100:.1f}% for 3rd class passengers —
  a gap of {(class_survival[1]-class_survival[3])*100:.1f} percentage points.
- Female passengers survived at a rate of {sex_survival['female']*100:.1f}%,
  dramatically higher than the {sex_survival['male']*100:.1f}% survival rate for male passengers.
- This suggests that both socioeconomic status (class) and 'women and children first'
  evacuation protocol were strong predictors of survival on the Titanic.
"""
print(summary)
```

```python
from groq import Groq
import getpass

api_key = getpass.getpass("Enter your Groq API key: ")
client = Groq(api_key=api_key)
```

```python
prompt = f"""
You are a data analyst. Write a short, professional 3-4 sentence summary of these findings
for a business presentation:

- 1st class survival rate: {class_survival[1]*100:.1f}%
- 2nd class survival rate: {class_survival[2]*100:.1f}%
- 3rd class survival rate: {class_survival[3]*100:.1f}%
- Female survival rate: {sex_survival['female']*100:.1f}%
- Male survival rate: {sex_survival['male']*100:.1f}%
"""
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[{"role": "user", "content": prompt}]
)

print(response.choices[0].message.content)
```
